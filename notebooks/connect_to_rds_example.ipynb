# Step-by-Step Guide to Using Database Functions
This notebook demonstrates how to use the provided functions to connect to an RDS PostgreSQL database, create a schema, create a table, and insert data from a DataFrame.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent

sys.path.append(str(ROOT / "src"))

from pathlib import Path

print("cwd =", Path.cwd())
print("parent =", Path.cwd().parent)
```

```python
# Import necessary modules
import pandas as pd

from aws_auth import get_secret
from config import (
    DATABASE_NAME,
    HOST_ENDPOINT,
    PORT,
    PROFILE_NAME,
    REGION_NAME,
    SECRET_NAME,
)
from rds_postgres_connection import (
    connect_to_rds,
    create_schema,
    create_table,
    insert_data,
    insert_data_bulk,
)
```

## Step 1: Retrieve Database Credentials
Use the `get_secret` function to retrieve the database credentials from AWS Secrets Manager.

```python
# Retrieve credentials from AWS Secrets Manager
credentials = get_secret(SECRET_NAME, region_name=REGION_NAME, profile_name=PROFILE_NAME)
print("PROFILE_NAME =", PROFILE_NAME)
print("REGION_NAME =", REGION_NAME)
print("SECRET_NAME =", SECRET_NAME)
if credentials is None:
    raise ValueError(
        "Failed to retrieve database credentials from AWS Secrets Manager."
    )
user = credentials["username"]
password = credentials["password"]
```

## Step 2: Connect to the Database
Use the `connect_to_rds` function to establish a connection to the database.

```python
# Connect to the RDS PostgreSQL database
conn = connect_to_rds(HOST_ENDPOINT, DATABASE_NAME, user, password, PORT)
```

## Step 3: Define a Sample DataFrame
Create a sample DataFrame to be inserted into the database.

```python
# Define a sample DataFrame
data = {
    "id": [1, 2, 3],
    "name": ["Alice", "Bob", "Charlie"],
    "age": [25, 30, 35]
}
df = pd.DataFrame(data)
df
```

## Step 4: Create a Schema
Use the `create_schema` function to create a schema in the database.

```python
# Define schema name
SCHEMA_NAME = "example"

# Create schema
create_schema(conn, SCHEMA_NAME)
```

## Step 5: Create a Table
Use the `create_table` function to create a table in the schema based on the DataFrame structure.

```python
# Define table name
TABLE_NAME = "sample_table"

# Create table
create_table(conn, df, SCHEMA_NAME, TABLE_NAME)
```

## Step 6: Insert Data
Use the `insert_data` function to insert the DataFrame into the table. If you have a lot of data to insert, use `insert_data_bulk` function instead.

```python
# Insert data into the table
insert_data(conn, df, SCHEMA_NAME, TABLE_NAME)
```

```python
insert_data_bulk(conn, df, SCHEMA_NAME, TABLE_NAME)
```

## Step 7: Close the Connection
Always close the database connection after completing the operations.

```python
# Close the connection
conn.close()
```
